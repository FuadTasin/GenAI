```python
def square(number):
    return number**2

print(square(4))
```

```python
square=lambda number:number**2

print(square(2))
```

```python
def is_adult(age):
    return age>=18

print(is_adult(19))
```

```python
is_age=lambda age: age>=18

print(is_age(2))
```

```python
x={"question":"What is 2+3?"}
```

```python
def check_question(x):
    for ch in x["question"]:
        if ch.isdigit():
            return True
    return False

check_question(x)        
```

```python
def check_question(x):
    return any(ch.isdigit() for ch in x["question"])

check_question(x)
```

```python
check_question=lambda x:any(ch.isdigit() for ch in x["question"])

check_question(x)
```
